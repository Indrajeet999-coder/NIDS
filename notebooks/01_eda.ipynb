```python
from pathlib import Path
import pandas as pd

raw_dir = Path("../dataset/raw/MachineLearningCVE")
csv_files = sorted(raw_dir.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files:\n")
for f in csv_files:
    size_mb = f.stat().st_size / (1024 * 1024)
    print(f"{f.name:<55} | {size_mb:.2f} MB")
```

```python
pd.set_option("display.max_columns", None)

sample_df = pd.read_csv(csv_files[0], nrows=100)
sample_df.columns = sample_df.columns.str.strip()

print("Shape:", sample_df.shape)
display(sample_df.head(5))
```

```python
label_counts = []

for f in csv_files:
    df_labels = pd.read_csv(f, usecols=lambda col: col.strip() == "Label")
    df_labels.columns = df_labels.columns.str.strip()
    counts = df_labels["Label"].value_counts()
    counts.name = f.stem
    label_counts.append(counts)

summary_table = pd.concat(label_counts, axis=1).fillna(0).astype(int)
summary_table["Total"] = summary_table.sum(axis=1)
display(summary_table)
```
