```python
from pathlib import Path
import pandas as pd

raw_dir = Path("../dataset/raw/MachineLearningCVE")
csv_files = sorted(raw_dir.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files:\n")
for f in csv_files:
    size_mb = f.stat().st_size / (1024 * 1024)
    print(f"{f.name:<55} | {size_mb:.2f} MB")
```

```python
pd.set_option("display.max_columns", None)

sample_df = pd.read_csv(csv_files[0], nrows=100)
sample_df.columns = sample_df.columns.str.strip()

print("Shape:", sample_df.shape)
display(sample_df.head(10))
```

```python
label_counts = []

for f in csv_files:
    df_labels = pd.read_csv(f, usecols=lambda col: col.strip() == "Label")
    df_labels.columns = df_labels.columns.str.strip()
    counts = df_labels["Label"].value_counts()
    counts.name = f.stem
    label_counts.append(counts)

summary_table = pd.concat(label_counts, axis=1).fillna(0).astype(int)
summary_table["Total"] = summary_table.sum(axis=1)
display(summary_table)
```

```python
import numpy as np

# Load a 100,000-row sample from Friday DDoS/PortScan file to inspect data quality
fri_df = pd.read_csv(csv_files[-1], nrows=100000)
fri_df.columns = fri_df.columns.str.strip()

inf_count = np.isinf(fri_df.select_dtypes(include=[np.number])).sum().sum()
nan_count = fri_df.isna().sum().sum()

print(f"Sample rows checked : {len(fri_df):,}")
print(f"Infinity values     : {inf_count:,}")
print(f"Missing (NaN) values: {nan_count:,}")
display(fri_df[["Destination Port", "Flow Duration", "Total Fwd Packets", "Average Packet Size", "Label"]].head(10))
```

```python
# Using the summary_table from Cell 3 across all 2.83 million rows
total_benign = summary_table.loc["BENIGN", "Total"]
total_attack = summary_table["Total"].sum() - total_benign
total_all = summary_table["Total"].sum()

print(f"Total Raw Flows       : {total_all:,}")
print(f"Normal (0 - BENIGN)   : {total_benign:,} ({total_benign/total_all*100:.2f}%)")
print(f"Intrusion (1 - Attack): {total_attack:,} ({total_attack/total_all*100:.2f}%)")
print(f"Imbalance Ratio       : {total_benign/total_attack:.2f} : 1")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
bars = plt.bar(["Normal (0)", "Intrusion (1)"], [total_benign, total_attack], color=["#1f4e78", "#c0392b"], width=0.45)
plt.title("CICIDS2017 Full Dataset Class Distribution (2.83 Million Flows)")
plt.ylabel("Number of Network Flows")
plt.ylim(0, total_benign * 1.15)
for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, h + 30000, f"{h:,}", ha="center", fontweight="bold")
plt.show()
```
